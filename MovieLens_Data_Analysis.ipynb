**2. DATA CLEANING PART**

2.1 Create copies of original datasets

```python
movies_clean = movies.copy()
ratings_clean = ratings.copy()
tags_clean = tags.copy()
links_clean = links.copy()

print("Dataset copies created successfully!")
```

2.2 Check and remove duplicate records

```python
print("Before Removing Duplicates")

print("Movies:", movies_clean.duplicated().sum())
print("Ratings:", ratings_clean.duplicated().sum())
print("Tags:", tags_clean.duplicated().sum())
print("Links:", links_clean.duplicated().sum())
```

Remove exact duplicate rows

```python
movies_clean = movies_clean.drop_duplicates()
ratings_clean = ratings_clean.drop_duplicates()
tags_clean = tags_clean.drop_duplicates()
links_clean = links_clean.drop_duplicates()

print("Duplicate records removed!")
```

2.3 Handle missing values

First, check missing values

```python
print("Missing Values Before Cleaning")

print("Movies:\n", movies_clean.isnull().sum())
print("Ratings:\n", ratings_clean.isnull().sum())
print("Tags:\n", tags_clean.isnull().sum())
print("Links:\n", links_clean.isnull().sum())
```

Handle missing values

```python
movies_clean = movies_clean.dropna(
    subset=['movieId', 'title']
)

ratings_clean = ratings_clean.dropna(
    subset=['userId', 'movieId', 'rating']
)

tags_clean = tags_clean.dropna(
    subset=['userId', 'movieId', 'tag']
)
```

2.4 Convert data types

```python
movies_clean['movieId'] = movies_clean['movieId'].astype('int64')

ratings_clean['userId'] = ratings_clean['userId'].astype('int64')
ratings_clean['movieId'] = ratings_clean['movieId'].astype('int64')

tags_clean['userId'] = tags_clean['userId'].astype('int64')
tags_clean['movieId'] = tags_clean['movieId'].astype('int64')

print("Data types converted successfully!")
```

2.5 Convert timestamp into datetime

```python
ratings_clean['timestamp'] = pd.to_datetime(
    ratings_clean['timestamp'], unit='s'
)

tags_clean['timestamp'] = pd.to_datetime(
    tags_clean['timestamp'], unit='s'
)

print("Timestamp converted successfully!")
```

2.6 Validate ratings

Movie ratings should be between 0.5 and 5.0, in increments of 0.5.

```python
ratings_clean = ratings_clean[
    (ratings_clean['rating'] >= 0.5) &
    (ratings_clean['rating'] <= 5.0) &
    ((ratings_clean['rating'] * 2) % 1 == 0)
]

print("Invalid ratings checked!")
```

2.7 Clean text columns

```python
movies_clean['title'] = movies_clean['title'].str.strip()

movies_clean['genres'] = movies_clean['genres'].str.strip()

tags_clean['tag'] = tags_clean['tag'].str.strip()

print("Text cleaning completed!")
```

2.8 Final verification

```python
print("FINAL DATASET SHAPES")

print("Movies:", movies_clean.shape)
print("Ratings:", ratings_clean.shape)
print("Tags:", tags_clean.shape)
print("Links:", links_clean.shape)

print("\nFINAL MISSING VALUES")

print("Movies:\n", movies_clean.isnull().sum())
print("Ratings:\n", ratings_clean.isnull().sum())
print("Tags:\n", tags_clean.isnull().sum())
print("Links:\n", links_clean.isnull().sum())
```

2.9 Save cleaned datasets

```python
movies_clean.to_csv('cleaned_movies.csv', index=False)
ratings_clean.to_csv('cleaned_ratings.csv', index=False)
tags_clean.to_csv('cleaned_tags.csv', index=False)
links_clean.to_csv('cleaned_links.csv', index=False)

print("All cleaned datasets saved successfully!")
```
